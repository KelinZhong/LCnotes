# Core Patterns

Roughly 80% of interview problems. Ordered easiest first — if you are working through this book linearly, this is the section that matters most.

| Notebook | Kind | Core idea |
| :--- | :--- | :--- |
| `06_arrays_and_hashing` | Reference | Preprocess once so range and lookup queries are O(1) |
| `07_two_pointers_and_sliding_window` | Study Guide | Move indices instead of rescanning |
| `08_binary_search` | Cheat Sheet | Halve a monotonic search space — over indices or over answers |
| `09_heap_and_monotonic_stack` | Cheat Sheet | Track the extreme element, globally (heap) or by position (stack) |
| `10_linked_lists` | Reference | Rewire pointers in place |
| `11_trees_and_graphs` | Study Guide | DFS and BFS, and everything built on them |
| `12_backtracking` | Cheat Sheet | Choose, explore, un-choose — when you must *list* every solution |
| `13_greedy` | Cheat Sheet | Take the locally best choice, when that is provably safe |
| `14_dynamic_programming` | Study Guide | Cache overlapping subproblems — the hardest topic here, deliberately last |

**If you do not know which of these a problem needs**, go back to `01_decision_guide` rather than guessing from the titles.
