# Python Toolkit & Complexity

The language-level material everything else assumes. Not topics in their own right — open them alongside whatever pattern you are working on.

| Notebook | Kind | Open it when |
| :--- | :--- | :--- |
| `02_python_builtins` | Reference | You need a method name, argument order, or an idiom for list / dict / set / deque |
| `03_python_strings` | Reference | Same, for strings |
| `04_complexity_reference` | Reference | You need the cost of an operation, or a memory footprint |
| `05_complexity_analysis` | Method | You need to *derive* the complexity of code you just wrote |

The two complexity notebooks answer different questions. `04` is a lookup table — what does `heapify` cost, how big is a list of a million ints. `05` is the method — recursion trees, amortised reasoning, and why a DFS is O(h) space rather than O(1).
