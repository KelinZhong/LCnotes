# Start Here

## If you have a problem in front of you

Open `01_decision_guide`. It routes an unseen problem to a pattern from its constraints and wording. Every other notebook here assumes you already know the topic — this is the only one that does not.

## If you are learning a topic

Work through **Core Patterns** in order. File numbers follow the suggested learning order, easiest first, with Dynamic Programming deliberately last. **Python Toolkit & Complexity** is language-level material to keep open alongside whatever you are working on. **Advanced** is lower-frequency and can wait.

## The three kinds of notebook

Every notebook carries one of these labels in italics under its title.

| Label | What it does | When to open it |
| :--- | :--- | :--- |
| **Study Guide** | Teaches the concept from scratch with explanations, comparisons and worked examples | Learning a topic for the first time — read it linearly |
| **Cheat Sheet** | Assumes you know the topic; gives the template, decision rules and common mistakes | Reviewing before an interview |
| **Reference** | Flat lookup, no prose | Mid-problem, when you need one specific trick or method |

Each group has its own contents page listing what is in it — use the sidebar, or the four group pages in the top navigation.
