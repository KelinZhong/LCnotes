# Decision Guide — Unseen Problem to Pattern

*Start here. Open this before you know what kind of problem you are looking at.*

Every other notebook assumes you have already named the topic. This one does not. It is keyed on what you can observe in the first thirty seconds — the constraints, the wording, and the shape of the input — so you can route a problem you have never seen to the notebook that solves it.

**How to use it:** read the constraints first (they eliminate the most), then scan the phrase table, then confirm with the input-shape table. Three passes, under a minute.

---

## The 30-second routine

1. **Read the constraint on `n`.** It tells you the complexity you are allowed, which usually eliminates most of the pattern space before you have thought about the problem at all.
2. **Underline the operative phrase.** "kth largest", "next greater", "number of ways", "all possible" — these map almost one-to-one onto patterns.
3. **Name the input shape.** Sorted array, grid, tree, graph, string, intervals. Each shape has a short list of candidates.
4. **State the brute force out loud, then ask what it repeats.** Recomputed subproblems point to DP or memoization; a rescanned window points to sliding window or prefix sums; a repeated max/min lookup points to a heap or monotonic stack.

---
## Table 1 — Constraints to complexity to pattern

Work backwards: the constraint fixes the complexity you are allowed, and the complexity fixes the short list of patterns.

| Constraint on `n` | Complexity you can afford | Patterns that fit |
| :--- | :--- | :--- |
| `n ≤ 12` | O(n!) | Permutation backtracking, brute-force ordering |
| `n ≤ 20` | O(2ⁿ) | Subset backtracking, bitmask enumeration, bitmask DP |
| `n ≤ 100` | O(n³) | Interval DP, Floyd-Warshall, triple loops |
| `n ≤ 1,000` | O(n²) | 2D DP, string DP, pairwise scans, LIS (quadratic form) |
| `n ≤ 10⁵` | O(n log n) | Sorting, binary search, heap, monotonic stack, sweep line |
| `n ≤ 10⁶` | O(n) | Single pass, prefix sums, two pointers, sliding window, counting |
| `n ≤ 10⁹` | O(log n) or O(1) | Binary search on the *answer*, math, closed form, bit tricks |

Two readings that carry most of the weight:

- **`n ≤ 20` almost always means exponential is intended.** Small bounds are permission, not a warning. If you see 20 and are hunting for a polynomial solution, you are probably overthinking it — go to `12_backtracking` or the bitmask section of `17_bit_manipulation`.
- **`n` up to 10⁹ cannot be iterated at all**, so the answer is never "scan the input". Either the input is small and the *value range* is huge (binary search on the answer — see `08_binary_search`), or there is a formula.

**A caution on multiple inputs.** When there are two bounds — say `n ≤ 10⁵` nodes and `m ≤ 10⁵` edges, or `n` strings of length `k` — the target complexity is usually in terms of the product or sum, not the larger one alone. Check that O(n · k) is actually affordable before committing to it.

**Phrases that mislead.** "Maximum" alone means nothing — it appears in greedy, DP, heap and binary-search-on-answer problems. What disambiguates it is whether choices are *independent* (greedy), *dependent on earlier choices* (DP), or *checkable given a candidate value* (binary search on answer).

## Table 4 — Disambiguating the near neighbours

These are the pairs that get confused most often. Each row is the question that separates them.

| Confusable pair | The question that decides |
| :--- | :--- |
| Greedy vs DP | Does a locally optimal choice stay optimal? If you can construct a case where taking the best move now loses later, it is DP. |
| Backtracking vs DP | Do you need to *output* every solution (backtracking), or just *count / optimise* over them (DP)? |
| Sliding window vs two pointers | Is the region between the pointers meaningful as a unit (window), or are you just converging from both ends (two pointers)? |
| Sliding window vs prefix sum | Are all values non-negative so shrinking is monotonic (window), or can values be negative (prefix sum + hash map)? |
| BFS vs DFS | Shortest path or level-order → BFS. Path existence, connectivity, or exploring to a leaf → DFS. |
| Union Find vs DFS | Are the edges arriving incrementally / do you need repeated "same group?" queries (Union Find), or is the graph fixed and you traverse once (DFS)? |
| Heap vs sorting | Do you need only the top k, or a running top element as data arrives (heap), or the whole ordering (sort)? |
| Binary search on index vs on answer | Are you searching *inside* an array (index), or over a range of candidate answer *values* (answer)? |
| Monotonic stack vs heap | Do you need the nearest qualifying element by position (stack), or the extreme element overall (heap)? |
| Memoised recursion vs tabulation | Same complexity — choose recursion when the state space is sparse or the transitions are awkward, tabulation when you want to drop the recursion stack. |

## When nothing matches

If three passes have not produced a candidate, work these in order:

1. **Write the brute force.** Even a hopeless one. Naming its complexity tells you the factor you need to remove, and the factor points at the tool: an inner rescan → prefix sum or sliding window; a repeated sort → heap; a recomputed subproblem → memoisation; a repeated "is this reachable" → Union Find.
2. **Sort the input and look again.** A surprising number of unsorted-array problems become two-pointer or greedy problems once sorted. Only viable if order is not part of the answer.
3. **Invert the question.** Counting what fails is sometimes far easier than counting what succeeds. "At most k" is often "at most k" minus "at most k−1".
4. **Reverse the direction.** Iterating from the right, or from the target back to the source, collapses some problems entirely — suffix maximums, reverse topological order, working backwards from the goal state.
5. **Look for the invariant.** Ask what stays true after every operation. Monotonic structures, XOR cancellation and cycle detection all fall out of an invariant rather than a search.
6. **Reduce to a known problem.** Most interview problems are a light disguise over one of about twenty classics. "Find the duplicate number" is cycle detection; "trapping rain water" is prefix max plus suffix max; "course schedule" is topological sort.
